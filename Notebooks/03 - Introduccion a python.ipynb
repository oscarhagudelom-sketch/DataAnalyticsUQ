Tipos de variables en python

1. Variables tipo string

```python
nombre="Oscar"
print(nombre)
print(type(nombre))
```

2. Variable tipo numero entero

```python
edad=20
print(edad)
print(type(edad))
```

3. Variable tipo numero decimal

```python
estatura=1.75
print(estatura)
print(type(estatura))
```

4. Variable tipo Booleano

```python
es_mayor_edad=True
es_menor_edad=False
print(es_mayor_edad)
print(es_menor_edad)
print(type(es_mayor_edad))
print(type(es_menor_edad))
```

```python
#suma
print(2+5)
#resta
print(2-5)
#multiplicacion
print(2*5)
#division
print(2/5)
#modulo
print(2%5)
#potencia
print(2**5)
```
